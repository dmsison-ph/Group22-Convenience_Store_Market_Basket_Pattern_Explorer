```python
import pandas as pd

from google.colab import files

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import (
    apriori,
    association_rules
)
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)
```

```python
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

df.head(50)
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate Transaction IDs:")
print(df["Transaction_ID"].duplicated().sum())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)
```

```python
transactions = (
    df["Purchased_Items"]
    .str.split(",")
    .apply(lambda items: [item.strip() for item in items])
    .tolist()
)

transactions[:5]
```

```python
te = TransactionEncoder()

encoded_array = te.fit(transactions).transform(transactions)

basket_df = pd.DataFrame(
    encoded_array,
    columns=te.columns_
)

basket_df.head()
```

```python
print("Number of transactions:", basket_df.shape[0])
print("Number of unique products:", basket_df.shape[1])

print("\nProducts:")
print(basket_df.columns.tolist())
```

```python
frequent_itemsets = apriori(
    basket_df,
    min_support=0.10,
    use_colnames=True,
    max_len=3
)

frequent_itemsets
```

```python
frequent_itemsets["support_count"] = (
    frequent_itemsets["support"] * len(df)
).round().astype(int)

frequent_itemsets = frequent_itemsets.sort_values(
    by="support_count",
    ascending=False
)

frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.80
)

rules
```

```python
rules_results = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

rules_results["support_count"] = (
    rules_results["support"] * len(df)
).round().astype(int)

rules_results = rules_results.sort_values(
    by="lift",
    ascending=False
)

rules_results
```

```python
def format_itemset(itemset):
    return ", ".join(sorted(itemset))

display_rules = rules_results.copy()

display_rules["antecedents"] = (
    display_rules["antecedents"].apply(format_itemset)
)

display_rules["consequents"] = (
    display_rules["consequents"].apply(format_itemset)
)

display_rules
```

```python
itemset_comparison = frequent_itemsets.copy()

itemset_comparison["itemsets"] = (
    itemset_comparison["itemsets"].apply(format_itemset)
)

itemset_comparison = itemset_comparison[
    ["itemsets", "support", "support_count"]
]

itemset_comparison
```

```python
display_rules.to_csv(
    "association_rules_results.csv",
    index=False
)

itemset_comparison.to_csv(
    "frequent_itemsets_results.csv",
    index=False
)

print("Results exported successfully.")
```
